```python
import pandas as pd
import numpy as np
from functools import reduce
```

$$
RV_t = \sum_{i=1}^{M} r_{t,i}^2
$$

where
- $t$ denotes the trading day  
- $i = 1, 2, \ldots, M$ indexes each intraday interval (here, 5-minute intervals)  
- $r_{t,i}$ is the continuously compounded intraday return  

Each 5-minute return is computed using closing prices as:

$$
r_{t,i} = \ln\left( \frac{P_{t,i}}{P_{t,i-1}} \right)
$$

where $P_{t,i}$ represents the closing price at the end of the $i$-th 5-minute interval.

```python
df = pd.read_csv("SPY.csv", skiprows=1, header=None, encoding='latin1',
                 names=['time','open','high','low','close','volume'])

df['time'] = pd.to_datetime(df['time'])
df = df.sort_values('time')
df['date'] = df['time'].dt.date
df['hour'] = df['time'].dt.hour
df['minute'] = df['time'].dt.minute

df = df[((df['hour'] > 9) | ((df['hour'] == 9) & (df['minute'] >= 30))) &
        (df['hour'] < 16)]

df['log_ret'] = np.log(df['close'] / df['close'].shift(1))
df.loc[df['date'] != df['date'].shift(1), 'log_ret'] = np.nan

rv = (df.groupby('date')['log_ret'].apply(lambda x: np.sum(x**2)).reset_index(name='RV'))
rv.to_csv("SPY_daily_RV.csv", index=False)
print(rv.head())
print(f"{len(rv)} days realized volatility")
```

```python
start_date = "2018-12-03"
end_date   = "2024-01-30"

etfs = ["SPY", "XLE", "XLF", "XLK", "XLY"]
rv_dfs = []

for etf in etfs:
    df = pd.read_csv(f"{etf}_daily_RV.csv")
    df['Date'] = pd.to_datetime(df['date'], errors='coerce')
    df = df[(df['Date'] >= start_date) & (df['Date'] <= end_date)]
    df = df[['Date', 'RV']] if 'RV' in df.columns else df[['Date', df.columns[-1]]]
    df = df.rename(columns={'RV': f'{etf}_RV'})
    rv_dfs.append(df)

rv_all = reduce(lambda left, right: pd.merge(left, right, on='Date', how='inner'), rv_dfs)
rv_all = rv_all.sort_values('Date').reset_index(drop=True)
```

```python
epu = pd.read_csv("EPU_Daily.csv")
epu['Date'] = pd.to_datetime(dict(year=epu['year'], month=epu['month'], day=epu['day']))
epu = epu[['Date', 'daily_policy_index']].rename(columns={'daily_policy_index': 'EPU'})
epu = epu[(epu['Date'] >= start_date) & (epu['Date'] <= end_date)]
print(epu.head())
```

```python
gpr = pd.read_excel("GPR_Daily.xls")
gpr = gpr[['date', 'GPRD']].rename(columns={'date': 'Date', 'GPRD': 'GPR'})
gpr['Date'] = pd.to_datetime(gpr['Date'])
gpr = gpr[(gpr['Date'] >= start_date) & (gpr['Date'] <= end_date)]
print(gpr.head())
```

```python
tpu = pd.read_excel("TPU_Daily.xlsx", sheet_name="TPU_DAILY")
tpu = tpu[['DATE', 'TPUD_index']].rename(columns={'DATE': 'Date', 'TPUD_index': 'TPU'})
tpu['Date'] = pd.to_datetime(tpu['Date'])
tpu = tpu[(tpu['Date'] >= start_date) & (tpu['Date'] <= end_date)]
print(tpu.head())
```

```python
fnsi = pd.read_excel("FNSI_daily.xlsx", sheet_name="Data")
fnsi = fnsi[['date', 'News Sentiment']].rename(columns={'date': 'Date', 'News Sentiment': 'FNSI'})
fnsi['Date'] = pd.to_datetime(fnsi['Date'])
fnsi = fnsi[(fnsi['Date'] >= start_date) & (fnsi['Date'] <= end_date)]
print(fnsi.head())
```

```python
dataset = (rv_all
            .merge(epu, on='Date', how='left')
            .merge(gpr, on='Date', how='left')
            .merge(tpu, on='Date', how='left')
            .merge(fnsi, on='Date', how='left'))

dataset = dataset.sort_values('Date').reset_index(drop=True)
dataset.to_csv("dataset.csv", index=False)
print(dataset.head())
```

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
import pandas as pd

cols = ["EPU", "GPR", "TPU", "FNSI"]

scaler = StandardScaler()
standardized = scaler.fit_transform(dataset[cols])
standardized_df = pd.DataFrame(standardized, columns=cols)
standardized_df["Date"] = dataset["Date"]

plt.figure(figsize=(12, 6))
plt.plot(standardized_df["Date"], standardized_df[cols])
plt.xlabel("Date")
plt.ylabel("Standardized Value (Z-score)")
plt.title("Standardized Trends of EPU, GPR, TPU, FNSI")
plt.legend(cols)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
